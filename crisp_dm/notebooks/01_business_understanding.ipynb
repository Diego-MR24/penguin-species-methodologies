#### Situación

Una estación de investigación del archipiélago Palmer registra cada temporada las medidas de los pingüinos que captura. La especie se anota a mano en campo, y un error de captura pasa a la base de datos sin que nadie lo note.

#### Objetivo del negocio

Revisar automáticamente la especie anotada en cada registro y marcar para revisión los registros en los que el modelo no coincide o no tiene suficiente confianza.

#### Objetivo de minería de datos

Clasificar la especie (Adelie, Chinstrap o Gentoo) a partir de las cuatro medidas, la isla y el sexo, con una máquina de vectores de soporte (SVM).

#### Recursos y restricciones

- **Datos:** `penguins_size.csv`, 344 registros de 2007 a 2009.
- **Herramientas:** Python con pandas y scikit-learn.
- **Restricción:** el modelo solo aplica a las tres islas de la base de datos.

#### Librerías

```python
import json
from pathlib import Path
```

#### Rutas

Relativas a la carpeta del notebook.

```python
INTERIM_DIR = Path("../data/interim")
INTERIM_DIR.mkdir(parents=True, exist_ok=True)
```

#### Criterios de éxito

- F1 macro en prueba de al menos 0.95.
- Sensibilidad de al menos 0.90 en cada especie, incluida la menos frecuente.
- Predicciones con probabilidad menor que 0.80 marcadas para revisión manual.

```python
success_criteria = {
    "f1_macro_min": 0.95,
    "recall_min_per_species": 0.90,
    "review_threshold": 0.80,
}
success_criteria
```
